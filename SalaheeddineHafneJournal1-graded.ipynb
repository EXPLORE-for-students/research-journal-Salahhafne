### Key Equations

The gravitational potential $\Phi_b(r)$ is defined as the work required to move a unit mass from a radius $r$ to infinity. Based on Equation (9) of our research , we calculated this by integrating the baryonic acceleration:
$$\Phi_b(r) = -\int_{r}^{\infty} \frac{V_b^2(r')}{r'} dr'$$
The total baryonic velocity squared, $V_b^2$, is the sum of the contributions from the stellar disk, the bulge, and the gas, weighted by their respective mass-to-light ratios ($\Upsilon$):
$$V_b^2 = \Upsilon_{disk} V_{disk}^2 + \Upsilon_{bulge} V_{bulge}^2 + V_{gas}^2$$
To account for the mass beyond our maximum observed radius ($R_{max}$), we apply a "tail correction" assuming a Keplerian decline:

$$\Phi_b(r) = -\left( \int_{r}^{R_{max}} \frac{V_b^2(r')}{r'} dr' + V_b^2(R_{max}) \right)$$

    def solve_baryonic_potential(r, v_disk, v_bulge, v_gas, upsilon_d=0.5, upsilon_b=0.7):
    v_b_sq = (upsilon_d * v_disk**2) + (upsilon_b * v_bulge**2) + v_gas**2
    integrand = v_b_sq / r
    
    phi_b = np.zeros_like(r)
    
    # integration for rs
    for i in range(len(r)):
        # r[i] to Rmax
        internal_integral = trapz(integrand[i:], r[i:])
        
         
        tail_correction = v_b_sq[-1] 
        
        phi_b[i] = -(internal_integral + tail_correction)
        
    return phi_b
